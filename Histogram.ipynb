### Write a Python program using OpenCV to perform histogram equalization on the given grayscale image **"parrot.jpg"**.

The program should:

- Import the required libraries (**OpenCV, NumPy, Matplotlib**).
- Read the image **"parrot.jpg"** in **grayscale** format.
- Display the grayscale image and plot its histogram.
- Perform **histogram equalization** on the grayscale image using `cv2.equalizeHist()` to enhance its contrast.
- Using Matplotlib, display the following in a **2 × 2 grid layout**:
  - Original Grayscale Image
  - Histogram of Original Image
  - Enhanced (Equalized) Image
  - Histogram of Enhanced Image


```python
# Import required libraries
import cv2
import matplotlib.pyplot as plt
import numpy as np



```

```python
# Read the image in grayscale format
image = cv2.imread('parrot.jpg')
gray_image = cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
```

```python
# Display the grayscale image.
plt.imshow(gray_image,cmap = 'gray')
plt.axis('off')
plt.show()
```

```python
# Plot the histogram of the grayscale image
plt.title("GrayScale Image Histogram")
histo = plt.hist(gray_image.ravel(),256,range = [0,256])
```

```python
# Perform histogram equalization
equalized_image = cv2.equalizeHist(gray_image)
plt.imshow(equalized_image)
plt.axis('off')
plt.show()




```

```python
# Display [1] the Original Image (Gray Image) and its Histogram, and [2] the Enhanced Image and its Histogram using a 2×2 layout in Matplotlib.
plt.title("Original Image")
plt.imshow(gray_image)
plt.axis('off')
plt.show()

plt.title("Histogram for the Original Image")
histo = plt.hist(gray_image.ravel(), 256, range=[0, 256])
plt.show()

plt.title("Enhanced Image")
plt.imshow(equalized_image)
plt.axis('off')
plt.show()

plt.title("Histogram for Enhanced Image")
histo2 = plt.hist(equalized_image.ravel(), 256, range=[0, 256])
```

### Write a Python program using OpenCV to perform color image enhancement through histogram equalization in the HSV color space on the given image **"parrot.jpg"**.

The program should:

- Import the required libraries (**OpenCV, NumPy, Matplotlib**).
- Read the color image **"parrot.jpg"**.
- Plot the **histogram of the original color image** (use the **B, G, and R** channels).
- Convert the image from **BGR to HSV** color space.
- Apply **histogram equalization** to one specific channel to improve brightness and contrast using `cv2.equalizeHist()`.
- Convert the enhanced HSV image back to **BGR** format.
- Using Matplotlib, display the following in a **2 × 2 grid layout**:
  - Original Color Image
  - Histogram of Original Image (BGR channels)
  - Enhanced Color Image
  - Histogram of Enhanced Image (BGR channels)


```python
# Read the color image
plt.title("Color Image")
plt.imshow(image)
plt.axis('off')
plt.show()



```

```python
# Convert BGR image to RGB
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
# Create histogram
plt.figure(figsize=(10, 5))
hist_blue = cv2.calcHist([image], [0], None, [256], [0, 256])
plt.plot(hist_blue, color='b', label='Blue')
hist_green = cv2.calcHist([image], [1], None, [256], [0, 256])
plt.plot(hist_green, color='g', label='Green')
hist_red = cv2.calcHist([image], [2], None, [256], [0, 256])
plt.plot(hist_red, color='r', label='Red')
plt.title("Histogram of Color Image")
plt.xlim([0, 256])
plt.legend()
plt.grid(alpha=0.2)
plt.show()
```

```python
# Convert to HSV.
hsv_image = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
```

```python
# Perform histogram equalization
h, s, v = cv2.split(hsv_image)
v_eq = cv2.equalizeHist(v)
hsv_eq = cv2.merge([h, s, v_eq])
```

```python
# Convert back to BGR format
enhanced_bgr = cv2.cvtColor(hsv_eq, cv2.COLOR_HSV2BGR)


```

```python
# Display
enhanced_rgb = cv2.cvtColor(enhanced_bgr, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(10, 8))

# 1. Original Color Image (reusing image_rgb from Cell 9)
plt.subplot(2, 2, 1)
plt.imshow(image_rgb)
plt.title("Original Color Image")
plt.axis('off')

# 2. Histogram of Original Image (BGR)
plt.subplot(2, 2, 2)
plt.plot(cv2.calcHist([image], [0], None, [256], [0, 256]), color='b')
plt.plot(cv2.calcHist([image], [1], None, [256], [0, 256]), color='g')
plt.plot(cv2.calcHist([image], [2], None, [256], [0, 256]), color='r')
plt.title("Histogram of Original Image (BGR)")
plt.xlim([0, 256])

# 3. Enhanced Color Image
plt.subplot(2, 2, 3)
plt.imshow(enhanced_rgb)
plt.title("Enhanced Color Image")
plt.axis('off')

# 4. Histogram of Enhanced Image (BGR)
plt.subplot(2, 2, 4)
plt.plot(cv2.calcHist([enhanced_bgr], [0], None, [256], [0, 256]), color='b')
plt.plot(cv2.calcHist([enhanced_bgr], [1], None, [256], [0, 256]), color='g')
plt.plot(cv2.calcHist([enhanced_bgr], [2], None, [256], [0, 256]), color='r')
plt.title("Histogram of Enhanced Image (BGR)")
plt.xlim([0, 256])

plt.tight_layout()
plt.show()



```
